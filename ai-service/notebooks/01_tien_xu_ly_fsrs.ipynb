```python
import random, re
from tqdm import tqdm
from huggingface_hub import list_repo_files

files = list_repo_files(REPO, repo_type="dataset", token=tok)
csv_files = [f for f in files if f.endswith('.csv')]
print(f"Tong file CSV: {len(csv_files)}")

random.seed(42)
chosen = random.sample(csv_files, 300)

dfs = []
for f in tqdm(chosen):
    try:
        path = hf_hub_download(REPO, f, repo_type="dataset", token=tok)
        d = pd.read_csv(path)
        d['user_id'] = int(re.search(r'/(\d+)\.csv$', f).group(1))
        dfs.append(d)
    except Exception as e:
        print(f"Bo qua {f}: {e}")

df = pd.concat(dfs, ignore_index=True)
print(f"\n{len(df):,} ban ghi tu {df.user_id.nunique()} nguoi dung")

df.to_parquet('/kaggle/working/raw_300users.parquet')
print("Da luu raw_300users.parquet")
```

```python
import pandas as pd
df = pd.read_parquet('/kaggle/working/raw_300users.parquet')
print(f"{len(df):,} dòng | {df.user_id.nunique()} user")
```

```python
print(df.state.value_counts())
print(df.rating.value_counts(normalize=True).round(3))
```

```python
c = df.sort_values(['user_id','card_id','review_th']).reset_index(drop=True)
before = len(c)

c = c[c.rating.between(1, 4)]
c['is_first'] = ~c.duplicated(['user_id','card_id'])

# lượt đầu (elapsed_days = -1): GIỮ, làm mốc khởi tạo
# lượt sau: chỉ giữ nếu cách >= 1 ngày (bỏ learning-step trong ngày), <= 10 năm
c = c[c.is_first | c.elapsed_days.between(1, 3650)].reset_index(drop=True)

c['i'] = c.groupby(['user_id','card_id']).cumcount() + 1      # thứ tự lượt trong thẻ
c['delta_t'] = c.elapsed_days.where(c.i > 1, 0)               # lượt đầu -> 0
c['y'] = (c.rating >= 2).astype(int)                          # 1 = nhớ, 0 = quên

cnt = c.groupby(['user_id','card_id'])['i'].transform('size')
c = c[cnt >= 2].reset_index(drop=True)

print(f"{before:,} -> {len(c):,} dòng ({len(c)/before:.1%} giữ lại)")
print("user:", c.user_id.nunique(), "| thẻ:", c.groupby(['user_id','card_id']).ngroups)
print("Tỉ lệ nhớ (chỉ lượt i>1):", round(c.loc[c.i > 1, 'y'].mean(), 3))
c.to_parquet('/kaggle/working/clean_300users.parquet')
```

```python
print(c.state.value_counts(normalize=True).round(4))
```

```python
import numpy as np

c = c.sort_values(['user_id','card_id','i']).reset_index(drop=True)
g = c.groupby(['user_id','card_id'])

c['n_reviews'] = c['i'] - 1                          # số lần ôn TRƯỚC lượt này
c['lapse']     = (c.y == 0).astype(int)
c['n_lapses']  = g['lapse'].cumsum() - c['lapse']    # số lần quên TRƯỚC lượt này

# ---- chia tập THEO NGƯỜI DÙNG ----
users = np.array(sorted(c.user_id.unique()))
rng = np.random.default_rng(42)
rng.shuffle(users)
n_test = int(len(users) * 0.2)
test_users, train_users = set(users[:n_test]), set(users[n_test:])

train = c[c.user_id.isin(train_users)].reset_index(drop=True)
test  = c[c.user_id.isin(test_users)].reset_index(drop=True)

# ---- assert chống rò rỉ ----
assert not (train_users & test_users), "RÒ RỈ: user xuất hiện ở cả hai tập!"
assert not (set(train.user_id) & set(test.user_id)), "RÒ RỈ ở mức dữ liệu!"
print(f"train: {len(train):,} dòng / {train.user_id.nunique()} user")
print(f"test : {len(test):,} dòng / {test.user_id.nunique()} user")
print("Tỉ lệ nhớ train/test:",
      round(train.loc[train.i>1,'y'].mean(),3), "/", round(test.loc[test.i>1,'y'].mean(),3))

train.to_parquet('/kaggle/working/train.parquet')
test.to_parquet('/kaggle/working/test.parquet')
```
